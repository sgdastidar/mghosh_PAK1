```python
import pandas as pd
import numpy as np
```

```python

# ===== INPUT =====
input_file = "dio_PAKPV_whole_dataset.csv"        # your DIO csv file
output_file = "feature_importance_PAKPV.txt"
```

```python
# ===== LOAD DATA =====
df = pd.read_csv(input_file)
print(df)
```

```python
# If first column is frame index, drop it
if not np.issubdtype(df.iloc[:,0].dtype, np.number):
    df = df.iloc[:,1:]

# ===== TAKE ABSOLUTE VALUE =====
abs_df = df.abs()

# ===== FILTER: keep values > 0 =====
# (threshold can be changed if needed)
threshold = 0.00
```

```python
# ===== AGGREGATE IMPORTANCE =====
# Take mean importance across frames (you can also use max)
feature_importance = abs_df.mean(axis=0)

# ===== FILTER FEATURES > threshold =====
feature_importance = feature_importance[feature_importance > threshold]

# ===== SORT DESCENDING =====
feature_importance = feature_importance.sort_values(ascending=False)

# ===== SAVE OUTPUT =====
feature_importance.to_csv(output_file, sep='\t', header=["Importance"])

print(feature_importance)
```
